```python
#lab 6

!pip install mlxtend

import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori

transactions = [
    ["Paracetamol","Vitamin C","Antibiotic"],
    ["Paracetamol","Vitamin C","Pain Reliever"],
    ["Paracetamol","Vitamin C","Antibiotic","Pain Reliever"],
    ["Paracetamol","Vitamin C","Antibiotic"],
    ["Vitamin C","Pain Reliever","Antacid"],
    ["Paracetamol","Vitamin C","ORS"],
    ["Paracetamol","Antibiotic","Pain Reliever"],
    ["Paracetamol","Vitamin C","Antibiotic","ORS"],
    ["Vitamin C","Pain Reliever","Calcium"],
    ["Paracetamol","Vitamin C","Antibiotic"],

    ["Paracetamol","Vitamin C","Antibiotic","Pain Reliever"],
    ["Paracetamol","Vitamin C","Multivitamin"],
    ["Vitamin C","Antibiotic","Pain Reliever"],
    ["Paracetamol","Vitamin C","Antibiotic","Antacid"],
    ["Paracetamol","Vitamin C","Pain Reliever"],
    ["Vitamin C","ORS","Calcium"],
    ["Paracetamol","Vitamin C","Antibiotic","Pain Reliever"],
    ["Paracetamol","Vitamin C","Antibiotic"],
    ["Paracetamol","Pain Reliever","Antacid"],
    ["Paracetamol","Vitamin C","Antibiotic","Calcium"],

    ["Vitamin C","Pain Reliever","Multivitamin"],
    ["Paracetamol","Vitamin C","ORS"],
    ["Paracetamol","Vitamin C","Antibiotic","Pain Reliever"],
    ["Paracetamol","Vitamin C","Antibiotic"],
    ["Paracetamol","Vitamin C","Pain Reliever"],
    ["Vitamin C","Antibiotic","ORS"],
    ["Paracetamol","Vitamin C","Antibiotic","Pain Reliever"],
    ["Paracetamol","Vitamin C","Antacid","Cough Syrup"],
    ["Vitamin C","Pain Reliever","Calcium","Cough Syrup"],
    ["Paracetamol","Vitamin C","Antibiotic","Pain Reliever","Cough Syrup"]
]

df = pd.DataFrame({
    "Transaction_ID": [f"T{i+1}" for i in range(len(transactions))],
    "Medicines": transactions
})

display(df)
```

```python
num_transactions = len(transactions)

unique_items = sorted(set(item for transaction in transactions for item in transaction))

print("Domain Selected :", "Hospital Pharmacy Medicine Analysis")
print("Number of Transactions :", num_transactions)
print("Number of Unique Items :", len(unique_items))

print("\nUnique Medicines")

for item in unique_items:
    print(item)
```

```python
te = TransactionEncoder()
te_array = te.fit(transactions).transform(transactions)

transaction_df = pd.DataFrame(te_array, columns=te.columns_)

display(transaction_df.head())

frequent_itemsets = apriori(
    transaction_df,
    min_support=0.30,
    use_colnames=True
)

frequent_itemsets["Support Count"] = (
    frequent_itemsets["support"] * len(transactions)
).astype(int)

frequent_itemsets = frequent_itemsets.sort_values(
    by="support",
    ascending=False
)

print("Minimum Support : 30%")
print("\nFrequent Itemsets")

display(frequent_itemsets)
```

```python
fused_patterns = []

itemsets = frequent_itemsets["itemsets"].tolist()

for i in range(len(itemsets)):
    for j in range(i + 1, len(itemsets)):
        common = itemsets[i].intersection(itemsets[j])

        if len(common) >= 1:
            merged = itemsets[i].union(itemsets[j])

            if merged not in fused_patterns:
                fused_patterns.append(merged)

fusion_df = pd.DataFrame({
    "Fused Pattern": [", ".join(sorted(pattern)) for pattern in fused_patterns],
    "Number of Items": [len(pattern) for pattern in fused_patterns]
})

print("Pattern Fusion")

display(fusion_df)
```

```python
largest_size = max(len(pattern) for pattern in fused_patterns)

colossal_patterns = [
    pattern
    for pattern in fused_patterns
    if len(pattern) == largest_size
]

support_values = []

for pattern in colossal_patterns:

    count = sum(
        pattern.issubset(set(transaction))
        for transaction in transactions
    )

    support_values.append(count)

colossal_df = pd.DataFrame({
    "Colossal Pattern":[", ".join(sorted(pattern)) for pattern in colossal_patterns],
    "Support Count":support_values,
    "Number of Items":[largest_size]*len(colossal_patterns)
})

print("Colossal Patterns")

display(colossal_df)
```

```python
highest_support = frequent_itemsets.iloc[0]

print("TASK 5 ANALYSIS\n")

print("Pattern with Highest Support")
print(highest_support["itemsets"])
print("Support Count :", highest_support["Support Count"])

print("\nLargest Colossal Pattern")

for pattern in colossal_patterns:
    print(pattern)

print("\nNumber of Fusion Operations :", len(fused_patterns))

print("\nFrequent Itemsets Before Fusion :", len(frequent_itemsets))

print("Patterns After Fusion :", len(fused_patterns))

print("\nPattern Fusion reduces redundancy by merging frequent patterns that share common items into larger meaningful patterns instead of storing many overlapping patterns.")
```
