```python
# ==========================================
# Question 1
# Hospital Pharmacy Analysis using Apriori
# ==========================================

import pandas as pd
from itertools import combinations

# Transactions
transactions = [
    ['Paracetamol','Vitamin C','Cough Syrup'],
    ['Paracetamol','Antibiotic','Vitamin C'],
    ['Antibiotic','Vitamin C','Pain Reliever'],
    ['Paracetamol','Vitamin C','Pain Reliever'],
    ['Paracetamol','Antibiotic','Vitamin C'],
    ['Vitamin C','Cough Syrup']
]

min_support = 3

# -----------------------
# Candidate 1 Itemsets
# -----------------------

C1 = {}

for transaction in transactions:
    for item in transaction:
        C1[item] = C1.get(item,0)+1

print("\nCandidate Itemsets C1")
print(pd.DataFrame(C1.items(),columns=['Medicine','Support Count']))

# Frequent 1-itemsets

L1 = {k:v for k,v in C1.items() if v>=min_support}

print("\nFrequent Itemsets L1")
print(pd.DataFrame(L1.items(),columns=['Medicine','Support Count']))

# -----------------------
# Candidate 2 Itemsets
# -----------------------

C2={}

for transaction in transactions:
    for pair in combinations(sorted(transaction),2):
        C2[pair]=C2.get(pair,0)+1

print("\nCandidate Itemsets C2")
print(pd.DataFrame(C2.items(),columns=['Itemset','Support Count']))

L2={k:v for k,v in C2.items() if v>=min_support}

print("\nFrequent Itemsets L2")
print(pd.DataFrame(L2.items(),columns=['Itemset','Support Count']))

# -----------------------
# Candidate 3 Itemsets
# -----------------------

C3={}

for transaction in transactions:
    for triple in combinations(sorted(transaction),3):
        C3[triple]=C3.get(triple,0)+1

print("\nCandidate Itemsets C3")
print(pd.DataFrame(C3.items(),columns=['Itemset','Support Count']))

L3={k:v for k,v in C3.items() if v>=min_support}

print("\nFrequent Itemsets L3")
print(pd.DataFrame(L3.items(),columns=['Itemset','Support Count']))

print("\n================================")
print("Medicine Stock Recommendation")
print("================================")

for item in L2:
    print(item)
```
