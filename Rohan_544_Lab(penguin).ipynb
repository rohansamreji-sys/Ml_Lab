```python

# LAB EXERCISE 1: EXPLORATORY DATA ANALYSIS - PENGUINS DATASET

# Import Libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Load Dataset
df = sns.load_dataset('penguins')

# Display First 5 Rows
print("First 5 Records:")
display(df.head())


# PART A: DATASET UNDERSTANDING


print("\nPART A: DATASET UNDERSTANDING")

# 1. Domain
print("\n1. Domain:")
print("Wildlife Biology / Ecology")

# 2. Number of Records and Attributes
print("\n2. Dataset Shape:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# 3. Features and Data Types
print("\n3. Features and Data Types:")
print(df.dtypes)

# 4. Target Variable
target = 'species'
print("\n4. Target Variable:")
print(target)

# 5. Purpose of Dataset
print("\n5. Dataset Purpose:")
print("To study physical characteristics of penguins and classify species.")


# PART B: DATA QUALITY ASSESSMENT


print("\n\nPART B: DATA QUALITY ASSESSMENT")

# 1. Missing Values
print("\n1. Missing Values:")
print(df.isnull().sum())

# Missing Values Heatmap
plt.figure(figsize=(8,5))
sns.heatmap(df.isnull(), cmap='viridis', cbar=False)
plt.title("Missing Values Heatmap")
plt.show()

# 2. Duplicate Records
print("\n2. Duplicate Records:")
print(df.duplicated().sum())

# 3. Inconsistent Entries
print("\n3. Unique Values in Categorical Columns:")
for col in df.select_dtypes(include='object'):
    print(f"\n{col}:")
    print(df[col].unique())

# 4. Data Completeness
print("\n4. Data Completeness (%)")
completeness = (1 - df.isnull().sum()/len(df))*100
print(completeness)

# 5. Handling Missing Values
df_clean = df.copy()

for col in df_clean.select_dtypes(include=np.number):
    df_clean[col].fillna(df_clean[col].median(), inplace=True)

for col in df_clean.select_dtypes(include='object'):
    df_clean[col].fillna(df_clean[col].mode()[0], inplace=True)

print("\nMissing Values After Cleaning:")
print(df_clean.isnull().sum())

# PART C: FEATURE ANALYSIS


print("\n\nPART C: FEATURE ANALYSIS")

# 1. Statistical Summary
print("\n1. Statistical Summary:")
display(df_clean.describe())

# 2. Feature Categorization
numerical_features = df_clean.select_dtypes(include=np.number).columns.tolist()
categorical_features = df_clean.select_dtypes(include='object').columns.tolist()

print("\n2. Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)

# 3. Histograms
df_clean.hist(figsize=(12,8))
plt.suptitle("Distribution of Numerical Features")
plt.tight_layout()
plt.show()

# 4. Correlation Matrix
plt.figure(figsize=(8,6))
sns.heatmap(df_clean[numerical_features].corr(),
            annot=True,
            cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()

# 5. Feature Relevance to Species
for col in numerical_features:
    plt.figure(figsize=(6,4))
    sns.boxplot(x='species', y=col, data=df_clean)
    plt.title(f"{col} vs Species")
    plt.show()

# Pairplot
sns.pairplot(df_clean, hue='species')
plt.show()

# Redundant Features
print("\nHighly Correlated Features:")
corr_matrix = df_clean[numerical_features].corr().abs()

for col in corr_matrix.columns:
    high_corr = corr_matrix[col][corr_matrix[col] > 0.80]
    if len(high_corr) > 1:
        print(f"\n{col}:")
        print(high_corr)

# Feature Engineering
df_clean['body_mass_kg'] = df_clean['body_mass_g'] / 1000

df_clean['bill_ratio'] = (
    df_clean['bill_length_mm'] /
    df_clean['bill_depth_mm']
)

print("\nNew Features Added:")
display(df_clean[['body_mass_kg','bill_ratio']].head())


# PART D: INSIGHT GENERATION


print("\n\nPART D: INSIGHT GENERATION")

# Species Distribution
plt.figure(figsize=(6,4))
sns.countplot(x='species', data=df_clean)
plt.title("Species Distribution")
plt.show()

# Island vs Species
plt.figure(figsize=(7,5))
sns.countplot(x='island', hue='species', data=df_clean)
plt.title("Species Distribution Across Islands")
plt.show()

# Body Mass Distribution
plt.figure(figsize=(6,4))
sns.boxplot(x='species', y='body_mass_g', data=df_clean)
plt.title("Body Mass by Species")
plt.show()


# FINAL INSIGHTS


print("\nFINAL INSIGHTS")

print("""
1. The dataset belongs to the Wildlife Biology domain.

2. The dataset contains information about three penguin species:
   - Adelie
   - Chinstrap
   - Gentoo

3. Missing values were found and handled using:
   - Median (Numerical Features)
   - Mode (Categorical Features)

4. No significant duplicate records were present.

5. Important features:
   - Flipper Length
   - Body Mass
   - Bill Length
   - Bill Depth

6. Strong correlations exist between:
   - Flipper Length and Body Mass

7. Species can be distinguished using physical measurements.

8. Feature engineering created:
   - body_mass_kg
   - bill_ratio

9. Potential Machine Learning Tasks:
   - Species Classification
   - Clustering
   - Feature Importance Analysis
   - Body Mass Prediction
""")
```
