# Lab Exercise 4 - FP Tree

```python
from collections import Counter

def frequency_table(transactions):
    f=Counter()
    for t in transactions: f.update(t)
    return f

def filter_sort_transactions(transactions,freq,minsup):
    order=sorted([i for i,c in freq.items() if c>=minsup], key=lambda x:(-freq[x],x))
    om={k:i for i,k in enumerate(order)}
    out=[]
    for t in transactions:
        items=[i for i in t if freq[i]>=minsup]
        items.sort(key=lambda x:om[x])
        out.append(items)
    return out

def build_tree(transactions):
    tree={}
    for t in transactions:
        node=tree
        for item in t:
            if item not in node:
                node[item]={"count":0,"children":{}}
            node[item]["count"]+=1
            node=node[item]["children"]
    return tree

def show(tree,ind=""):
    for k,v in tree.items():
        print(ind+"|--",k,"("+str(v["count"])+")")
        show(v["children"],ind+"    ")

def run(title,transactions,minsup):
    print("="*40);print(title);print("="*40)
    print("\nTransactions")
    for i,t in enumerate(transactions,1): print(f"T{i}: {t}")
    freq=frequency_table(transactions)
    print("\nFrequency Table")
    for k,v in freq.items(): print(k,":",v)
    ft=filter_sort_transactions(transactions,freq,minsup)
    print("\nSorted Transactions")
    for i,t in enumerate(ft,1): print(f"T{i}: {t}")
    tree=build_tree(ft)
    print("\nFP Tree")
    print("ROOT")
    show(tree)
movie=[["Action","Comedy","Thriller"],["Action","Drama","Thriller"],["Comedy","Thriller","Romance"],["Action","Comedy","Thriller"],["Action","Comedy","Drama"]]
smart=[["Smart Light","Smart TV","Air Conditioner"],["Smart Light","Smart Speaker","Air Conditioner"],["Smart TV","Air Conditioner","Smart Speaker"],["Smart Light","Smart TV","Air Conditioner"],["Smart Light","Air Conditioner","Smart Speaker"]]
run("Scenario 1",movie,3)
print()
run("Scenario 2",smart,3)
```
